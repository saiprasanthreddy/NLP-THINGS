Once you identify the text file(s) you want to work with (e.g., `your_text_file.txt`), you can read its content into a Python variable like this. Make sure to replace `your_text_file.txt` with the actual name of your file.

```python
import os

# Replace 'your_text_file.txt' with the actual path to your text file inside the unzipped directory
# Based on the `ls` output, a sample file is '/content/sentence_classification_data/SentenceCorpus/labeled_articles/arxiv_annotate10_7_1.txt'
file_path = '/content/sentence_classification_data/SentenceCorpus/labeled_articles/arxiv_annotate10_7_1.txt' # Updated this path

if os.path.exists(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        text_content = f.read()
    print(f"Successfully read file: {file_path}\n")
    print(" characters of the file:")
    print(text_content)
else:
    print(f"Error: File not found at {file_path}. Please check the correct file name and path from the `ls` output above.")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
sentance = nltk.sent_tokenize(text_content)
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
import re
```

```python
corpous = []
for i in range(len(sentance)):
  word = re.sub('[^a-zA-Z]', ' ', sentance[i])
  word = word.lower()
  word = word.split()

  word = [lemmatizer.lemmatize(word) for word in word if word not in set(stopwords.words('english'))]
  word = ' '.join(word)
  corpous.append(word)
```

```python
#  bag of words

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(binary = True,ngram_range=(3,3))
X = cv.fit_transform(corpous).toarray()

cv.vocabulary_
```
